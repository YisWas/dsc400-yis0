```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Year": [2020, 2021, 2022, 2023, 2024, 2025],
    "Celtics": [3, 4, 2, 2, 1, 1],
    "Bucks": [1, 2, 3, 1, 3, 4],
    "Heat": [4, 3, 1, 4, 5, 3],
    "76ers": [5, 1, 4, 3, 2, 5],
    "Knicks": [8, 7, 6, 5, 4, 2]
}

df = pd.DataFrame(data)

for team in df.columns[1:]:
    plt.plot(
        df["Year"],
        df[team],
        marker="o",
        linewidth=20,  # Makes the lines thicker
        label=team
    )

plt.gca().invert_yaxis()

plt.xlabel("Year")
plt.ylabel("Rank")
plt.grid()
plt.title("NBA Team Rankings Over Time")
plt.legend()

plt.show()
```
