```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer


sent_text = "There is a Fake News report out there that I am going to remove the White House Press Room and bring back the old Swimming Pool, which lies directly underneath. Although I may move the Fake News to a less obnoxious location, the old swimming pool will never be brought back. I NEVER even thought to do such a stupid thing! Thank you for your attention to this matter. President DJT"
sent_text_no_thank =  "There is a Fake News report out there that I am going to remove the White House Press Room and bring back the old Swimming Pool, which lies directly underneath. Although I may move the Fake News to a less obnoxious location, the old swimming pool will never be brought back. I NEVER even thought to do such a stupid thing! President DJT"
# Create sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Analyze sentiment
scores = sia.polarity_scores(sent_text)
scores_no_thank = sia.polarity_scores(sent_text_no_thank)
# Print results
print(scores)
print(scores_no_thank)
```

A compound score of **-0.91** indicates very negative sentiment, with -1 being the most negative.

The overall tone of the text is negative, even though the subject isn't inherently negative, such as war or disease.

Negative terms include:
1. Fake News
2. Obnoxious
3. NEVER (emphasizes rejection rather than being inherently negative)
4. Stupid

When I remove the word `thank`, the positive sentiment score drops to `pos: 0`, but that's a small .05 change, vs the overall neg score of .2 and it only changes the compound score by .02

```python
import spacy
from IPython.display import display, HTML
from spacy import displacy

nltk.download("vader_lexicon", quiet=True)
nlp = spacy.load("en_core_web_sm")


topic_text = """I am proud to announce the NEWEST FLAGSHIP NATIONAL PARK in the Country, CHIRICAHUA NATIONAL PARK in ARIZONA! In my First Term, I signed the GREAT American Outdoors Act, one of the LARGEST Investments in ... Congratulations to all of our other Great Congressmen in the State, including Andy Biggs, David Schweikert, Eli Crane, Abe Hamadeh, and Paul Gosar. We are MAKING OUR PARKS, AND ARIZONA, GREAT AGAIN! President DONALD J. TRUMP"""

doc = nlp(topic_text)

print("Sentences:", len(list(doc.sents)))
print('\nFirst sentence:\n')
print(list(doc.sents)[0],"\n\n")

for ent in doc.ents:
    print(f"{ent.text:<40} {ent.label_:<10} {spacy.explain(ent.label_)}")

displacy.render(doc, style="ent", jupyter=True)
```
