```python
import os
import requests
from dotenv import load_dotenv
import pandas as pd
```

```python
load_dotenv()

api_key = os.getenv("FRED_API_KEY")

url = "https://api.stlouisfed.org/fred/series/observations"
sticky_cpi_yoy = 'CORESTICKM159SFRBATL'
params = {
    "series_id": sticky_cpi_yoy,
    "api_key": api_key,
    "file_type": "json"
}

response = requests.get(url, params=params)
```

```python
response
```

```python
# A 200 status code means the request was successful
```

```python
# Extract observations and create the table
data = response.json()
data.keys()
```

```python
data
```

```python
cpi_table = pd.DataFrame(data["observations"])
```

```python
cpi_table.head()
```

```python
cpi_table.info()
```

```python
# This shows that there are no null values, but I noticed that the first value in cpi_table['value'] was "."
# I will convert each column to the appropriate data type and try again.
```

```python
# Drop the first row because its value is "."
cpi_table = cpi_table.drop(cpi_table.index[0])

# Convert columns
cpi_table["realtime_start"] = pd.to_datetime(cpi_table["realtime_start"])
cpi_table["realtime_end"] = pd.to_datetime(cpi_table["realtime_end"])
cpi_table["date"] = pd.to_datetime(cpi_table["date"])
cpi_table["value"] = pd.to_numeric(cpi_table["value"])
```

## Visualizations: Comparing Cooldowns

```python
# After each peak, let's compare the speed of the cooldown.
```

```python
# Let's view the whole dataset

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 2))

plt.plot(
    cpi_table["date"],
    cpi_table["value"]
)

plt.title("Sticky-Price Consumer Price Inflation Over Time")
plt.xlabel("Date")
plt.ylabel("Year-over-Year Inflation Rate (%)")
plt.grid(alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt

# I manually verified the peak dates for each target cycle.
# Each range covers the first 20 months after the peak so the comparison is consistent.
target_ranges = {
    "1970s": ("1975-04-01", "1976-12-01"),
    "1980s": ("1980-06-01", "1982-02-01"),
    "2020s": ("2022-12-01", "2024-08-01")
}

plt.figure(figsize=(10, 6))

for name, (start, end) in target_ranges.items():

    # Get only the data inside this range
    period = cpi_table[
        (cpi_table["date"] >= start) &
        (cpi_table["date"] <= end)
    ].copy()

    # Standardize the x-axis to start at 0 instead of using calendar dates
    period["month"] = range(len(period))

    plt.plot(period["month"], period["value"], label=name)

plt.xlabel("Months Since Peak")
plt.ylabel("Inflation Rate (%)")
plt.title("Sticky-Price Inflation: First 20 Months After Each Peak")
plt.legend()
plt.grid(True)

plt.show()
```

- The 2020s cycle peaked at the lowest inflation rate of the three cycles.
- Over the first 20 months after each peak, inflation declined by about 37% in the 2020s, compared with about 47% in the 1970s and 40% in the 1980s. This suggests that the most recent cooldown was somewhat more gradual.
